## <font size="6"> **Introduction** </font>

<font size="4"> The nuclear reactor, which is the heart of a nuclear power plant, generates thermal energy that is converted into electrical energy for use in the economy through various means. To prevent any unfortunate incidents, it is essential to continuously perform several key parameter calculations. These include calculations of multiplication factors, reactivity coefficients, fuel temperature (Doppler effect), the impact of neutron poisons on reactivity, burnup effects, isotopic concentration changes with burnup, fast and thermal neutron flux density, axial and radial power peaking of the core, fission rate distribution, and overall core power distribution. To ensure the integrity of the nuclear reactor core during operation, these calculations are regularly performed and evaluated.

<font size="4"> Some of the most well-known Monte Carlo neutron transport codes, such as MCNP or SERPENT are gaining popularity as leading tools for reactor core physics computations. Unfortunately, many of these codes, which are frequently used for neutronic calculations, are not easily accessible, and their distribution is often restricted. However, some reactor physics codes, such as the OpenMC Monte Carlo code and the deterministic DRAGON code, are freely available and increasingly used in code-to-code comparisons.

<font size="4"> In another study, OpenMC was used to investigate the "VVER-1000 MOX core computational benchmarks" and was successfully applied to the neutronics calculation of the entire VVER-1000 core. Various benchmark problems can be used to thoroughly assess the core of a VVER reactor. For this investigation, a VVER-440 full core containing low-enrichment assemblies (ranging from 1.6 to 4.25 wt%) was used as a benchmark problem.

<font size="4">The first reference study involves a 2D FULL-CORE benchmark, performed by ŠKODA JS research team, presented at the 21st Symposium of AER in 2011 where solution has been calculated by MCNP with ENDF/B6  nuclear data library. The second reference benchmark involves the same assumptions applied on DYN3D determinist code, based on libraries generated with Serpent on ENDF/VII.1 library. 

<font size="4">The benchmark problem specifies the various parameters and values to be calculated.

# <font size="5"> **A. Common assumptions** </font>

<font size="4"> The models are represented in OpenMC using Python (Python 3.7) code in this
 Notebook with the latest version of OpenMC. OpenMC features hexagonal geometry modeling, which was used to design each type of assembly separately.

<font size="4"> We focus on the active zone, taking into account the geometry of the explicit radial reflector.

<font size="4"> Different materials in various regions within the assemblies are defined using Boolean operations for modeling, also known as constructive solid geometry. To account for thermal scattering at lower energies, S(α,β) tables are provided. Additionally, the following modelings include cross section temperature interpolation using the OpenMC option, ensuring proper thermal treatment of the materials.


<font size="4"> It is assumed that the emission is isotropic, which simplifies the angular distribution of the emitted particles. Furthermore, the energy distribution is modeled by a Watt spectrum with specific shape and scale parameters.

<font size="4"> There are 349 assemblies. Each assembly contains 127 elementary cells of various types, such as differently enriched fuel pins, gadolinium pins, central tubes which are filled with moderator.

![Différents types de tubes guides](https://drive.google.com/uc?id=1rPElNynJZOH3-4lPbr8YQjmeKu5kLEUa)

<font size="4"> The 147 mm distance between the centers of two neighboring assemblies is maintained in every state geometry. The thickness of the ‘vygorodka’ (internal edged metal sheet) is 8 mm, and the gap between the ‘vygorodka’ and the FA model hexagon (with a pitch of 147 mm) is 1.5 mm (resulting in a 2.5 mm water gap between the ‘vygorodka’ and the border FA shroud). Fuel pins have an **active height assumed to be 250 cm** for 3D computations.

![Description de l'image](https://drive.google.com/uc?id=1rhC00gxJLTz-28vn5OIIw0ezvp962Ra0)

<font size="4"> The core temperature is set to 543.15 K and is uniform for all materials. Although all dimensions refer to the cold state geometry, the pressure is **12.3 MPa** and the corresponding moderator density is **0.777537 g/cm³**.

<font size="4"> Finally here other commun assumptions applicable for our two simulations :

* <font size="4"> Isothermal conditions.
* Full spectrum calculation.
* The boric acid concentration is fixed at 3 g/kg (525 ppm).
* 135I and 135Xe concentrations are set to 0.0.
* No thermal-hydraulic calculations.
* Temperature interpolation of ENDF cross sections using OpenMC.
* If 3D simulation, the reflection boundary conditions are used in axial directions.
* The steel tube between the “vygorodka” and the core
basket drawn in the picture is neglected in the calculations.
* Detailed fixed geometry as described  above.

# <font size="5"> **B. 2D FULLCORE benchmark** </font>

![Description de l'image](https://drive.google.com/uc?id=19Ku-CT0OEO9AvAbdT0YsHajoEUqrY-Cl)

<font size="4">The VVER-440 FULLCORE is a two-dimensional numerical problem based on the VVER-440 reactor core under cold conditions. There are three types of fuel assemblies with 1.6%, 2.4%, and 4.25% average 235U enrichment. The fuel assembly with the highest enrichment contains a burnable absorber in the form of gadolinium oxide mixed with uranium oxide. The content of gadolinium oxide in fuel pellets is 3.35 wt%. The central tubes of all FA’s are
filled with moderator.

<font size="4"> The reflection
boundary conditions are used on the azimuthal edges while the total absorption
boundary condition (leakage to the vacuum) is used on the cylindrical outer boundary. As 2D simulation, both the core and radial reflector structure in axial direction is treated like infinite homogeneous.

<font size="4"> The horizontal cut of the core may be covered by a net of hexagons and by this way so called reflector FA’s may be modeled. We'll take into account one row of “reflector” FAs in the calculation.

<font size="4">The parameter of interest calculated with OpenMC is the **assembly-wise and pin-wise power peaking factors**. The integral fission power from the FAs will be normalized in such a way that the mean value of all 349 FAs in 360-degree symmetry equals 1. The pin-by-pin power distribution will be normalized in such a way that the mean value of the relative pin powers equals 1 in each FA.

<font size="4">The main factors affecting the precision of the Monte Carlo power distribution calculation are the number of neutrons per generation, the number of generations, and the source locations. Reference calculations have been performed with the MCNP-4C code, mainly for the core assemblies periphery V&V purposes. The ENDF/B-VI library was chosen, with a temperature of 543.15 K using the tmp option and with thermalization lwtr.04t.

<font size="4">In our case with OpenMC, we will perform 200,000 neutrons per batch with 1,000 batches and account for thermal scattering with an S(α,β) table (as mentioned in common assumptions). Also, inactive batches have to be set in OpenMC; their number depends on how quickly the simulation reaches a steady state. To properly estimate them, we'll analyze the convergence of keff during a preliminary calculation.

```python
#================================================
# Materials definition
#==============================================

# --------------------------
# Zircalloy (E-110 et E-125)
# --------------------------

E110 = openmc.Material(name='E110')
E110.add_element('Zr', 0.9878, 'wo')
E110.add_element('Nb', 0.0100, 'wo')
E110.add_element('O', 0.001, 'wo')
E110.add_element('Fe', 0.0005, 'wo')
E110.add_element('C', 0.0002, 'wo')
E110.set_density('g/cm3', 6.55)

E125 = openmc.Material(name='E125')
E125.add_element('Zr', 0.9728, 'wo')
E125.add_element('Nb', 0.0250, 'wo')
E125.add_element('O', 0.0010, 'wo')
E125.add_element('Fe', 0.0005, 'wo')
E125.add_element('C', 0.0002, 'wo')
E125.set_density('g/cm3', 6.55)

# -------------------------------------------------------------
# Pressure vessel, core barrel, core basket, 'vygorodka' steel
# -------------------------------------------------------------
#steel = openmc.Material(name='Steel')
#steel.add_element('Fe', 0.68535, 'wo')
#steel.add_element('Cr', 0.18000, 'wo')
#steel.add_element('Ni', 0.10000, 'wo')
#steel.add_element('Mn', 0.02000, 'wo')
#steel.add_element('Si', 0.00800, 'wo')
#steel.add_element('Ti', 0.00550, 'wo')
#steel.add_element('C', 0.00080, 'wo')
#steel.add_element('P', 0.00035, 'wo')
#steel.set_density('g/cm3', 8.0)

# --------------------------
# Moderator (borated water)
# --------------------------
moderator = openmc.Material(name='Moderator')
moderator.add_element('H', 0.6665, 'ao')
moderator.add_element('O', 0.3332, 'ao')
moderator.add_nuclide('B10', 0.000058, 'ao')
moderator.add_nuclide('B11', 0.000234, 'ao')
moderator.add_s_alpha_beta('c_H_in_H2O') #thermal S(α,β) scattering option
moderator.set_density('g/cm3', 0.777537)

# --------------------------
# UO2 fuels
# --------------------------

atoms = {
    'O16':  [4.64026e22, 4.64069e22, 4.64132e22, 4.64153e22, 4.64174e22, 4.65970e22],
    'U234': [3.02003e18, 4.52999e18, 6.79486e18, 7.54980e18, 8.30473e18, 7.29688e18],
    'U235': [3.75892e20, 5.63831e20, 8.45731e20, 9.39696e20, 1.03366e21, 9.08216e20],
    'U238': [2.28224e22, 2.26351e22, 2.23541e22, 2.22604e22, 2.21668e22, 2.15147e22],
    'Gd154': [0.0000, 0.0000, 0.0000, 0.0000, 0.0000, 2.52388e19],
    'Gd155' : [0.0000, 0.0000, 0.0000, 0.0000, 0.0000, 1.71346e19],
    'Gd156' : [0.0000, 0.0000, 0.0000, 0.0000, 0.0000, 2.36990e20],
    'Gd157' : [0.0000, 0.0000, 0.0000, 0.0000, 0.0000, 1.81177e20],
    'Gd158' : [0.0000, 0.0000, 0.0000, 0.0000, 0.0000, 2,87583e20],   
}

fuels = {}
name_fuels = []

# Enrichments available
enrichments = ["1.6", "2.4", "3.6", "4.0", "4.4", "4.0Gd"]

for i, enrich in enumerate(enrichments):
    name = f'fuel_{enrich.replace(".", "p")}'
    name_fuels.append(name)

    mat = openmc.Material(name=name)    
    # Extract atomic density
    dens_dict = {nuc: atoms[nuc][i] for nuc in atoms}
    
    # Total density (atom/cm3)
    n_tot = sum(dens_dict.values())
    mat.set_density('atom/cm3', n_tot)

    # Add each nucleide in atomic fraction
    for nuc, n_i in dens_dict.items():
      frac_ao = 100.0 * n_i / n_tot
      mat.add_nuclide(nuc, frac_ao, 'ao')

    fuels[name] = mat

materials = openmc.Materials([E110,E125,moderator,fuels["fuel_4p0Gd"],fuels["fuel_4p0"],fuels["fuel_4p4"],fuels["fuel_3p6"],fuels["fuel_2p4"],fuels["fuel_1p6"]])
materials.export_to_xml()
```

```python
# ================================================
# Fuel pins geometry definition
# ================================================

def create_pin(fuel, cladding, mod, pitch):
    # Pellet
    pellet = openmc.ZCylinder(r=0.76/2)
    fuel_cell = openmc.Cell(fill=fuel, region=-pellet)

    # Gainage
    cladding_outer = openmc.ZCylinder(r=0.91/2)
    cladding_cell = openmc.Cell(fill=cladding, region=+pellet & -cladding_outer)

    # Modérateur
    mod_cell = openmc.Cell(name=str(fuel),fill=mod, region=+cladding_outer)

    return openmc.Universe(cells=[fuel_cell, cladding_cell, mod_cell])

# Création des types de crayons
pin_pitch = 1.23  # cm
pin_16 = create_pin(fuels["fuel_1p6"], E110, moderator, pin_pitch)
pin_24 = create_pin(fuels["fuel_2p4"], E110, moderator, pin_pitch)
pin_44 = create_pin(fuels["fuel_3p6"], E110, moderator, pin_pitch)
pin_40 = create_pin(fuels["fuel_4p0"], E110, moderator, pin_pitch)
pin_36 = create_pin(fuels["fuel_4p4"], E110, moderator, pin_pitch)
pin_gd = create_pin(fuels["fuel_4p0Gd"], E110, moderator, pin_pitch)
pin_guide = create_pin(moderator, E110, moderator, pin_pitch)
```

```python
# ================================================
# Fuel assembly definition
# ================================================
assembly_pitch = 14.7
assembly_edge = (assembly_pitch/2)*(sqrt(3)/2)

moderator_outer_cell = openmc.Cell(name='Coolant outer', fill=moderator)
moderator_universe = openmc.Universe(cells=[moderator_outer_cell])

outer_shroud_prism = openmc.model.HexagonalPrism(edge_length = assembly_edge*1.1, boundary_type='transmission')
iner_shroud_prism = openmc.model.HexagonalPrism(edge_length = assembly_edge*1.2, boundary_type='transmission')
shroud_region = + iner_shroud_prism & - outer_shroud_prism
shroud_cell = openmc.Cell(fill=E125, region= shroud_region)
n = 0

def create_assembly(name, pattern):
    assembly = openmc.HexLattice(name=name)
    assembly.center = (0, 0)
    assembly.pitch = [pin_pitch]
    assembly.outer = moderator_universe
    
    # Pins lattice
    assembly.universes = pattern
    
    # FA's hexagonal configuration
    hex_prism = openmc.model.HexagonalPrism(edge_length = assembly_edge*1.5, boundary_type='transmission')
    #shroud = openmc.ZCylinder(r=14.5/2)
    
    # Converting Prism to inside region
    assembly_region = -hex_prism
    
    # Cellule de l'assemblage
    assembly_cell = openmc.Cell(fill=assembly, region= assembly_region)
    
    return openmc.Universe(cells=[assembly_cell,shroud_cell])

pattern425 = [
    ([pin_36] + [pin_40] * 5) * 6,
    ([pin_gd] + [pin_44] * 4) * 6,
    ([pin_44]*4) * 6,
    ([pin_44]*3) * 6,
    ([pin_44]*2) * 6,
    [pin_44] * 6,
    [pin_guide]]

pattern24 = [
    [pin_24]*36,
    [pin_24]*30,
    [pin_24]*24,
    [pin_24]*18,
    [pin_24]*12,
    [pin_24]*6,
    [pin_guide]]

pattern16 = [
    [pin_16]*36,
    [pin_16]*30,
    [pin_16]*24,
    [pin_16]*18,
    [pin_16]*12,
    [pin_16]*6,
    [pin_guide]]

assembly16 = create_assembly("FA_1.6%", pattern16)
assembly24 = create_assembly("FA_2.4%", pattern24)
assembly425 = create_assembly("FA_4.25%", pattern425)
assemblyempty = create_assembly("FA_4.25%", pattern425)
```

```python
# ================================================
# Cœur complet
# ================================================

# Configuration hexagonale du cœur
core = openmc.HexLattice(name='Core')
core.center = (0, 0)
core.pitch = [assembly_pitch]
core.outer = moderator_universe  # Réflecteur
core.orientation = 'x' # orientation des hexagones pour construire la structure voulue
# Placement des assemblages pour un coeur  complet de réacteur VVER440
core.universes = [
    ([moderator_universe] * 4 + [assembly425] * 4 + [moderator_universe] * 3) * 6,
    ([moderator_universe] + [assembly16] * 2 + [assembly24] + [assembly425]
     + [assembly16] + [assembly425] + [assembly24]
     + [assembly16] * 2) * 6,
    ([assembly16] + [assembly425] * 2 + [assembly16] + [assembly24] * 2
     + [assembly16] + [assembly425] * 2) * 6,
    ([assembly24] *2 + [assembly16]
     + [assembly24] + [assembly16]
     + [assembly24] + [assembly16]
     + [assembly24] ) * 6,
    ([assembly24]+[assembly16] * 6 ) * 6,
    ([assembly24]+[assembly16] * 2 + [assembly24]
     + [assembly16] * 2) * 6,
    ([assembly16] * 5) * 6,
    ([assembly16]*2 + [assembly24]
     + [assembly16]) * 6,
    ([assembly24] + 2* [assembly16]) * 6,
    ([assembly16] * 2) * 6,
    [assembly16] * 6,
    [assembly24]
] # À compléter selon disposition réelle

# Enveloppe du cœur
core_edge = (assembly_pitch*11)

#core_prism = openmc.model.HexagonalPrism(edge_length = core_edge, boundary_type='reflective')
core_prism = openmc.ZCylinder(r=core_edge, boundary_type='reflective')
core_region = -core_prism
core_cell = openmc.Cell(fill=core, region = core_region)

root_universe = openmc.Universe(cells=[core_cell],universe_id=0, name='root universe')
geometry = openmc.Geometry(root_universe)
```
